# **11 — Software Design and Architecture**



**Document type:** Master topic checklist, four-week study plan, and table of contents  
**Audience:** Developers in any track (backend, frontend, mobile, AI, data, DevOps, security, embedded, robotics, etc.)  
**Purpose:** Identify what to study, in a sensible order, with traceability to the two selected Udemy courses and clear gaps for independent study. This is an index and checklist, not a textbook.  
**Learning horizon:** Complete the **four-week Software Design & Architecture module** within the broader **three-month universal software-engineering foundation program**. Revisit Intermediate and Advanced sections as you build and operate larger systems.  
**Language:** Use one programming language consistently for exercises; course code examples may use a different language.  

> **Scope:** Course completion is not equivalent to mastery. Learn the concept, implement a small example, review an existing design, explain trade-offs, and demonstrate the result in a working project. Some topics require supplementary resources because they are not explicitly covered by the selected course lessons.

## Learning-source legend

- **Course A:** [Software Architecture & Design — Learn IT](https://www.udemy.com/course/software-architecture-learnit/) — main source for code-level design and Week 1–2.
- **Course B:** [Software Architecture & Design of Modern Large Scale Systems](https://www.udemy.com/course/software-architecture-design-of-modern-large-scale-systems/) — main source for system-level architecture and Week 3–4.
- **Independent:** Supplementary reading, implementation, review, and project work. Do not assume a named Udemy course covers it in depth.

*Course lesson names below are navigational references to the four-week course plan. Udemy may update lecture names or ordering.*
