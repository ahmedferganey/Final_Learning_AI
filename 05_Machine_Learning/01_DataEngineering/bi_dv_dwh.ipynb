## 1. What is Data?

Data is any raw fact, figure, observation, or measurement that, on its own, has little meaning until it is processed, organized, or interpreted. It is the fundamental building block of information.

**Key distinctions:**
- **Data**: Raw, unprocessed facts (e.g., `42`, `"John"`, `2024-01-15`)
- **Information**: Data that has been processed and given context (e.g., "John sold 42 units on January 15, 2024")
- **Knowledge**: Information combined with experience/insight that enables decision-making (e.g., "John's sales spike on the 15th correlates with payday — we should schedule promotions around that date")

**Sources of data:**
- Transactional systems (POS, ERP, CRM)
- Sensors and IoT devices
- Web and app interactions (clickstreams, logs)
- Surveys and forms
- Social media
- Third-party/external datasets (government, market research)

**Why data matters:**
Data is the raw material for every analytical or BI process. Poor-quality data ("garbage in") leads to poor decisions ("garbage out"), which is why data quality — accuracy, completeness, consistency, timeliness, and validity — is a foundational concern before any analysis begins.

## 2. Types of Data

Data is generally classified into three categories based on how it is organized and stored.

### a) Structured Data
Highly organized data that fits neatly into rows and columns, typically stored in relational databases or spreadsheets.

- **Characteristics**: Fixed schema, predefined format, easily searchable using SQL
- **Examples**: Sales transaction tables, employee records, financial ledgers, inventory databases
- **Storage**: Relational Database Management Systems (RDBMS) like MySQL, PostgreSQL, SQL Server, Oracle
- **Pros**: Easy to query, analyze, and visualize; mature tooling
- **Cons**: Rigid structure makes it hard to accommodate new/unexpected data types

### b) Unstructured Data
Data with no predefined format or organization, making it harder to collect, process, and analyze using traditional methods.

- **Characteristics**: No fixed schema, often text-heavy or multimedia
- **Examples**: Emails, videos, images, audio recordings, social media posts, PDF documents, customer reviews
- **Storage**: Data lakes, NoSQL databases, object storage (e.g., Amazon S3)
- **Pros**: Captures rich, real-world complexity (sentiment, context, visuals)
- **Cons**: Requires advanced techniques (NLP, computer vision, ML) to extract value
- **Estimate**: Often cited as making up 80–90% of all organizational data

### c) Semi-Structured Data
Data that doesn't reside in a rigid relational structure but contains tags, markers, or metadata that separate elements and enforce hierarchies, making it easier to analyze than fully unstructured data.

- **Characteristics**: Self-describing structure, flexible schema
- **Examples**: JSON files, XML documents, log files, emails (headers are structured, body is unstructured), NoSQL databases like MongoDB
- **Storage**: NoSQL databases, JSON/XML stores
- **Pros**: Flexible and scalable; bridges the gap between structured and unstructured
- **Cons**: Still requires parsing logic; not as instantly query-able as structured data

| Aspect | Structured | Semi-Structured | Unstructured |
|---|---|---|---|
| Schema | Fixed | Flexible/self-describing | None |
| Storage | RDBMS | NoSQL, JSON/XML | Data lakes, file stores |
| Examples | SQL tables, Excel | JSON, XML, logs | Images, video, text |
| Query ease | Easy (SQL) | Moderate | Hard (needs ML/NLP) |
| % of data | ~10-20% | ~10% | ~80% |

## 3. What is Business Intelligence (BI)?

**Business Intelligence (BI)** refers to the technologies, processes, and practices used to collect, integrate, analyze, and present business data to support better business decision-making. BI focuses primarily on **understanding the past and present** — "what happened and why" — through reporting and visualization.

**Core components of BI:**
1. **Data Warehousing**: Centralized repositories that consolidate data from multiple sources (e.g., Snowflake, Redshift, BigQuery)
2. **ETL/ELT Processes**: Extract, Transform, Load (or Extract, Load, Transform) — pipelines that move and clean data from source systems into a warehouse
3. **Data Modeling**: Structuring data (e.g., star schemas, fact/dimension tables) for efficient querying
4. **Reporting & Dashboards**: Visual representations of KPIs and metrics (e.g., Power BI, Tableau, Looker, Qlik)
5. **Ad-hoc Querying**: Enabling business users to explore data without needing to write complex code

**Typical BI outputs:**
- Sales performance dashboards
- Financial reporting
- Customer segmentation reports
- Operational KPI trackers (e.g., inventory turnover, churn rate)

**Goal of BI**: To turn raw data into actionable, digestible insights that non-technical business stakeholders can use to make informed decisions quickly — often through self-service dashboards updated in near real-time.

## 4. BI vs Data Analytics vs Data Science

These three fields are closely related and often overlap, but they differ in scope, techniques, and the questions they answer.

| Dimension | Business Intelligence (BI) | Data Analytics | Data Science |
|---|---|---|---|
| **Primary Question** | "What happened?" | "Why did it happen? What should we do?" | "What will happen? What's the optimal action?" |
| **Time Orientation** | Past & present | Past & present (deeper dive) | Present & future |
| **Techniques** | Dashboards, reports, SQL queries | Statistical analysis, hypothesis testing, segmentation | Machine learning, predictive modeling, AI |
| **Tools** | Power BI, Tableau, Looker | Excel, SQL, Python/R (pandas, statistics) | Python/R, TensorFlow, Scikit-learn, Spark |
| **Output** | Dashboards, KPI reports | Insights, trend analysis, recommendations | Predictive models, algorithms, automated systems |
| **Skill Focus** | Data visualization, business acumen | Statistics, data wrangling, critical thinking | Programming, ML, advanced statistics, big data engineering |
| **Example** | "Show me last quarter's regional sales" | "Why did sales drop in the Northeast in Q3?" | "Predict which customers will churn next month" |

**How they connect:**
- BI is often the **foundation** — it tells you *what* is happening via dashboards.
- Data Analytics goes a layer deeper, asking *why*, using statistical rigor to explain patterns.
- Data Science extends this further into *prediction and automation*, using machine learning to forecast future outcomes or optimize decisions at scale.

Think of it as a maturity spectrum: **BI (descriptive) → Analytics (diagnostic/some predictive) → Data Science (predictive/prescriptive, automated).**

## 5. Types of Analytics

Analytics can be categorized into four types, often visualized as a maturity curve — each level adds more complexity and business value.

### a) Descriptive Analytics — "What happened?"
Summarizes historical data to understand what has occurred.
- **Techniques**: Reporting, dashboards, data aggregation, summary statistics (mean, median, totals)
- **Example**: "Total revenue last month was $2.3M, a 5% increase from the previous month."
- **Tools**: Excel, Power BI, Tableau
- **Value**: Low complexity, but essential foundation for all further analysis

### b) Diagnostic Analytics — "Why did it happen?"
Digs deeper into data to understand the root causes of past outcomes.
- **Techniques**: Drill-down analysis, correlation analysis, data mining, root-cause analysis
- **Example**: "Revenue increased 5% because of a successful holiday marketing campaign in the Northeast region."
- **Tools**: SQL, Excel pivot tables, BI drill-through reports
- **Value**: Moderate complexity; explains patterns identified in descriptive analytics

### c) Predictive Analytics — "What will happen?"
Uses historical data, statistical algorithms, and machine learning to forecast future outcomes.
- **Techniques**: Regression analysis, time-series forecasting, classification models, machine learning
- **Example**: "Based on current trends, we predict a 12% increase in customer churn next quarter."
- **Tools**: Python, R, Azure ML, IBM SPSS
- **Value**: Higher complexity; enables proactive planning

### d) Prescriptive Analytics — "What should we do about it?"
Recommends specific actions to take based on predictive insights, often using optimization and simulation techniques.
- **Techniques**: Optimization algorithms, simulation, decision trees, AI-driven recommendation engines
- **Example**: "To reduce predicted churn by 8%, offer a 15% loyalty discount to at-risk customers in segments A and B."
- **Tools**: Advanced ML platforms, optimization software, AI decision engines
- **Value**: Highest complexity and highest business value; directly informs action

**Visual summary (complexity vs. value):**

```
Value/Complexity
     ^
     |                                    Prescriptive
     |                          Predictive
     |              Diagnostic
     |   Descriptive
     +--------------------------------------------> 
             Increasing Analytical Maturity
```

## 6. Basic Excel Skills for Data Analysis

Excel remains one of the most widely used tools for data analysis, especially for beginners and business users. Core skills include:

### a) Data Entry & Formatting
- Understanding cell references (relative `A1`, absolute `$A$1`, mixed `$A1`/`A$1`)
- Formatting data types (dates, currency, percentages, text)
- Data validation (dropdown lists, restricting input types)

### b) Core Formulas & Functions
- **Basic math**: `SUM`, `AVERAGE`, `COUNT`, `COUNTA`, `MAX`, `MIN`
- **Logical functions**: `IF`, `AND`, `OR`, `IFS`, `IFERROR`
- **Lookup functions**: `VLOOKUP`, `HLOOKUP`, `INDEX`/`MATCH`, `XLOOKUP` (modern replacement)
- **Text functions**: `CONCATENATE`/`CONCAT`, `LEFT`, `RIGHT`, `MID`, `TRIM`, `TEXT`
- **Conditional aggregation**: `SUMIF`/`SUMIFS`, `COUNTIF`/`COUNTIFS`, `AVERAGEIF`/`AVERAGEIFS`

### c) Data Organization
- **Sorting and Filtering**: Arranging data and isolating subsets based on criteria
- **PivotTables**: Summarizing, grouping, and cross-tabulating large datasets dynamically
- **PivotCharts**: Visualizing pivot table data instantly

### d) Data Cleaning
- Removing duplicates (`Remove Duplicates` tool)
- Handling blanks/errors (`IFERROR`, `Find & Replace`)
- Text-to-Columns for splitting data
- Power Query for more advanced cleaning/transformation (import, merge, reshape data)

### e) Data Visualization
- Basic charts: bar, line, pie, scatter
- Conditional formatting (color scales, data bars, icon sets) to highlight patterns
- Sparklines for compact in-cell trend visualization

### f) Basic Analysis Tools
- **What-If Analysis**: Goal Seek, Data Tables, Scenario Manager
- **Basic statistical functions**: `STDEV`, `VAR`, `CORREL`, `TREND`
- **Data Analysis ToolPak**: Add-in for regression, histograms, ANOVA, and more

**Why Excel matters as a foundation:**
Even as organizations adopt advanced BI tools (Power BI, Tableau) and programming languages (Python, R), Excel remains the universal starting point — it builds intuition for tabular data structures, formulas, and basic statistical thinking that transfers directly into more advanced analytics tools.

## Summary: How These Pieces Fit Together

1. **Data** (raw facts) → comes in **structured, semi-structured, or unstructured** forms
2. That data is collected, warehoused, and visualized through **BI tools** to answer "what happened"
3. **Data Analytics** and **Data Science** build on BI foundations to explain *why* and predict *what's next*
4. The **four types of analytics** (descriptive → diagnostic → predictive → prescriptive) represent an increasing maturity curve of insight and action
5. **Excel** serves as the accessible entry point for practicing these concepts before moving to specialized BI/analytics/data science tools
