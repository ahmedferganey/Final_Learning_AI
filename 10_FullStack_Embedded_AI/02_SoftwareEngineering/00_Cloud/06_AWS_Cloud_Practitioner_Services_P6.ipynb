## 1. What Is Amazon OpenSearch Service?

Amazon OpenSearch Service is a fully managed AWS service used to search, analyze, and visualize large volumes of structured and unstructured data.

It is based on the open-source OpenSearch search and analytics engine.

AWS manages much of the underlying infrastructure, including:

- Cluster provisioning and operation.
- Infrastructure monitoring.
- Node replacement after failures.
- Scaling options.
- Automated snapshots and recovery capabilities.
- Security and encryption features.

OpenSearch is commonly used for application search, log analytics, monitoring, and vector search for AI applications.

### AWS Service Classification

| Category | Classification |
|---|---|
| Primary AWS Category | Analytics |
| Core Function | Search and data analytics |
| Database Capability | Document-oriented storage and retrieval |
| AI Capability | Vector and semantic search |
| RAG Integration | Vector database / retrieval engine |

> **Important:** OpenSearch is not a traditional relational database such as Amazon RDS. It is primarily a distributed search and analytics engine, although it can store JSON documents and vector embeddings.

## 2. Why Do We Need OpenSearch?

Imagine an e-commerce application containing millions of products.

A customer searches for:

```text
Wireless headphones with noise cancellation
```

The application needs to:

1. Find relevant products.

3. Filter products by price.

A traditional relational database can support some of these operations, but a dedicated search engine is designed specifically for complex search workloads.

OpenSearch provides:

- Full-text search.
- Relevance-based ranking.
- Filtering and sorting.
- Aggregations and analytics.
- Vector similarity search.
- Distributed processing across multiple nodes.

## 3. Core OpenSearch Concepts

Understanding the following components is essential before working with OpenSearch.

### 3.1 Domain

A domain represents a managed OpenSearch cluster.

It contains the infrastructure required to store, index, and search data.

Example:

```text
Domain: products-search

    |
    +--- Data Node 1
    |
    +--- Data Node 2
    |
    +--- Data Node 3
```

When creating a provisioned domain, you configure settings such as:

- OpenSearch engine version.
- Instance types.
- Number of nodes.
- Storage capacity.
- Network configuration.
- Security settings.

### 3.2 Index

An index is a logical collection of related documents.

For example:

```text
OpenSearch Domain
    |
    +--- products
    |
    +--- customers
    |
    +--- application-logs
```

An index is conceptually similar to a database table, although its data model and indexing behavior are different.

### 3.3 Document

A document is a JSON object stored inside an index.

Example:

```json
{
  "product_id": 101,
  "name": "Wireless Headphones",
  "category": "Electronics",
  "price": 150,
  "description": "Bluetooth headphones with noise cancellation"
}
```

Each document has an identifier.

### 3.4 Shards

A shard is a partition of an index.

OpenSearch distributes large indexes into shards to support distributed storage and processing.

Example:

```text
Index: products

    |
    +--- Primary Shard 1
    |
    +--- Primary Shard 2
    |
    +--- Primary Shard 3
```

Each shard holds a portion of the indexed data.

### 3.5 Replicas

A replica is a copy of a primary shard.

Replicas provide:

- High availability.
- Improved search capacity.
- Resilience against node failures.

Example:

```text
Node A
  |
  +--- Primary Shard 1

Node B
  |
  +--- Replica Shard 1
```

## 4. OpenSearch Search Capabilities

OpenSearch supports multiple search approaches.

### 4.1 Full-Text Search

Searches text within documents.

Example:

```text
Search:

"Wireless Headphones"
```

OpenSearch can return documents containing matching terms, ordered by relevance.

### 4.2 Keyword Search

Searches for specific terms or exact values.

Example:

```text
Category = Electronics
```

Useful for:

- Product categories.
- IDs.
- Status values.
- Exact filters.

### 4.3 Vector Search

Vector search retrieves documents based on numerical representations of meaning, called embeddings.

Example:

```text
User Query:

"How can I deploy a Python API on AWS?"
```

A semantic search system may retrieve documents about:

```text
FastAPI
Docker
Amazon ECS
Amazon EC2
Application Load Balancer
```

Even when the documents do not contain every word in the original query, their meanings may be relevant.

### 4.4 Hybrid Search

Hybrid search combines traditional keyword search with semantic search.

```text
                 User Query
                      |
           +----------+----------+
           |                     |
           v                     v
     Keyword Search        Vector Search
           |                     |
           +----------+----------+
                      |
                      v
               Combine Results
                      |
                      v
                Final Results
```

This is particularly useful in enterprise search and RAG applications.

---

## 5. OpenSearch as a Vector Database

OpenSearch can function as a vector database by storing embeddings and searching for similar vectors.

It supports vector search using `knn_vector` fields, similarity measures, and approximate nearest-neighbor algorithms such as HNSW.

### 5.1 What Is a Vector Embedding?

An embedding is a numerical representation of data.

For example:

```text
Original Text:

"Amazon EC2 is a virtual server service."
```

An embedding model converts the text into a vector:

```text
[0.12, -0.35, 0.82, 0.41, ...]
```

The actual vector may contain hundreds or thousands of dimensions.

Documents with similar meanings tend to have similar embeddings.

### 5.2 Vector Search Process

```text
User Question
      |
      v
Embedding Model
      |
      v
Query Vector
      |
      v
Amazon OpenSearch
      |
      v
Vector Similarity Search
      |
      v
Top-K Similar Documents
```

The embedding model generates numerical representations.

OpenSearch stores and searches them.

> **Important:** OpenSearch does not need to be the service that generates the embeddings. An embedding model from Amazon Bedrock, SageMaker AI, or another model provider can perform that task.

## 6. OpenSearch Integration with Amazon Bedrock

One of the most important AI use cases for OpenSearch is Retrieval-Augmented Generation (RAG).

RAG allows an LLM to generate answers using relevant information retrieved from an external knowledge base.

Amazon Bedrock Knowledge Bases supports OpenSearch as a vector-store option.

```text
                  DOCUMENT INGESTION

                     PDF Documents
                          |
                          v
                      Amazon S3
                          |
                          v
                 Extract and Chunk Text
                          |
                          v
                Bedrock Embedding Model
                          |
                          v
                    Vector Embeddings
                          |
                          v
                 Amazon OpenSearch
                     Vector Store


                  QUESTION ANSWERING

                     User Question
                          |
                          v
                Bedrock Embedding Model
                          |
                          v
                     Query Vector
                          |
                          v
                 Amazon OpenSearch
                          |
                          v
                 Relevant Documents
                          |
                          v
                   Amazon Bedrock
                       LLM
                          |
                          v
                    Final Answer
```

## 7. Amazon OpenSearch Service vs OpenSearch Serverless

AWS offers two main deployment approaches.

| Feature | Provisioned OpenSearch | OpenSearch Serverless |
|---|---|---|
| Main resource | Domain | Collection |
| Capacity | Configure and manage capacity | Automatically scales capacity |
| Cluster management | AWS manages infrastructure; you configure nodes and capacity | AWS manages underlying compute and capacity |
| Scaling | Configurable scaling | Automatic scaling |
| Search | Yes | Yes |
| Vector Search | Yes | Yes |
| Billing | Provisioned resources and storage, plus applicable charges | OpenSearch Compute Units and storage, plus applicable charges |

OpenSearch Serverless separates the application workload from underlying cluster and node management.

### 7.1 Serverless Collection Types

OpenSearch Serverless supports three principal collection types:

| Collection Type | Purpose |
|---|---|
| Search | Full-text application search |
| Time Series | Log analytics and time-based data |
| Vector Search | Embeddings and semantic search |

Collection-type availability and capabilities can vary by Serverless generation and configuration.

For a RAG application, a vector search collection is a common choice.

## 10. OpenSearch Security

OpenSearch supports several security mechanisms.

| Security Layer | Purpose |
|---|---|
| IAM | Controls AWS management and applicable data-plane permissions |
| Domain Access Policies | Controls access to OpenSearch domain APIs |
| Fine-Grained Access Control | Controls access to indexes, documents, and fields |
| VPC | Provides private network access to provisioned domains |
| TLS/HTTPS | Encrypts data in transit |
| AWS KMS | Supports encryption-key management |
| Encryption at Rest | Protects stored data |

For a private application, a common architecture is:

```text
                 AWS VPC
                    |
       +------------+------------+
       |                         |
       v                         v
   ECS / EC2              OpenSearch Domain
       |                         ^
       |                         |
       +------ Private HTTPS ----+
```

An OpenSearch domain placed inside a VPC is not directly accessible from a computer outside that VPC without appropriate private connectivity.

# Amazon Cloud formation 


## AWS CloudFormation

### What is CloudFormation?

AWS CloudFormation is an **Infrastructure as Code (IaC)** service that allows you to create, manage, and delete AWS resources automatically using YAML or JSON templates.

Instead of manually creating resources in the AWS Console, you define them in a template.

### How It Works

```text
CloudFormation Template
       |
       v
CloudFormation Stack
       |
       v
AWS Resources
   |
   +--- VPC
   +--- EC2
   +--- Security Group
   +--- S3
   +--- RDS
```

### Key Concepts

| Concept | Description |
|---|---|
| Template | YAML or JSON file defining AWS resources |
| Stack | Collection of resources managed together |
| Change Set | Preview of proposed changes before applying them |
| StackSets | Deploy stacks across multiple AWS accounts and Regions |
| Drift Detection | Detects when resources differ from their template configuration |

### Example: Create an S3 Bucket

```yaml
AWSTemplateFormatVersion: '2010-09-09'

Description: Create an S3 bucket

Resources:
  MyBucket:
    Type: AWS::S3::Bucket
    Properties:
      BucketName: my-example-bucket-12345
```

CloudFormation creates the specified S3 bucket when the stack is deployed, provided the bucket name is available.

### CloudFormation vs Terraform

| CloudFormation | Terraform |
|---|---|
| AWS-native service | Multi-cloud IaC tool |
| YAML / JSON | HCL |
| Manages AWS resources | Manages AWS and other providers |
| Uses Stacks | Uses State files |

> **Remember:** CloudFormation = Define infrastructure in code → Deploy a stack → AWS creates and manages the resources.

| Technique | Description |
|---|---|

# Retrieval-Augmented Generation




## 24. What Is RAG?

**Retrieval-Augmented Generation (RAG)** combines information retrieval with generative AI.

It allows an LLM to use relevant external information when answering questions.

### Without RAG

```text
User Question
     |
     v
LLM
     |
     v
Answer From Model Knowledge
```

### With RAG

```text
User Question
     |
     v
Retrieve Relevant Documents
     |
     v
Provide Context to LLM
     |
     v
Generate Grounded Answer
```

### Example — Company Knowledge Assistant

A company has internal documents:

```text
Quality Manual

Production SOPs

Maintenance Instructions

Safety Procedures
```

The user asks:

```text
"What is the procedure for machine startup?"
```

RAG retrieves relevant sections before generating an answer.

## 26. RAG vs Fine-Tuning

| RAG | Fine-Tuning |
|---|---|
| Retrieves external information | Updates model behavior through training |
| Useful for changing knowledge | Useful for adapting task behavior or style |
| Does not ordinarily change model weights | Changes model parameters |
| Requires retrieval infrastructure | Requires training data and compute |
| Can provide source references | Does not inherently provide source citations |

### Example

Use RAG when:

```text
Question:
"What is the latest company policy?"
```

Use fine-tuning when:

```text
Requirement:
"Make the model consistently follow
a specialized output style or task."
```

RAG and fine-tuning can also be combined.

## 34. Overfitting vs Underfitting

### Overfitting

The model learns the training data too closely and performs poorly on unseen data.

```text
Training Performance:
High

Testing Performance:
Low
```

### Underfitting

The model fails to learn enough from the training data.

```text
Training Performance:
Low

Testing Performance:
Low
```

### Generalization

A well-generalized model performs adequately on unseen data.

# 35. Step 31 — Update Application Code Later

CloudFormation owns the infrastructure and embeds the initial application source.

The Flask application is created by EC2 User Data.

If you modify the application manually on the instance, those changes are outside CloudFormation's template state.

If you update the embedded application code in the YAML, simply updating the stack does not necessarily rerun User Data on the existing EC2 instance.

For this lab, the initial deployment is enough.

For production development, separate the Flask source code from the infrastructure template and use a proper application deployment pipeline.

Possible production architecture:
